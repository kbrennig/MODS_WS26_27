## Einen ersten Überblick bekommen

## Spalten auswählen
Eine Spalte wählen wir aus wie einen Wert aus einem Dictionary, über ihren Namen. Heraus kommt eine **Series** (eine einzelne Spalte), die sich wie ein NumPy-Array verhält. Mehrere Spalten wählen wir mit einer **Liste** von Namen, also doppelten eckigen Klammern.

```python
display(data["season"])   # wählen Sie eine Spalte aus, hier "season"
print(data["temp"].max()) #Methoden auf Spalten, hier max() auf Spalte "temp"
data[["temp", "humidity", "count"]].head() # wählen Sie mehrere Spalten aus, hier "temp", "humidity" und "count"
```

**.value_counts()** zählt, wie oft jeder Wert in einer Spalte vorkommt. 

```python
data["weather"].value_counts()    # wie viele Stunden pro Wetterkategorie?
```

## Zeilen filtern
Hier kommen die Masken aus dem Python-Intro (Teil 5) zurück: Eine Bedingung auf eine Spalte ergibt eine Maske, und mit der Maske wählen wir Zeilen aus.

```python
display(data.iloc[0])   # erste Zeile

maske_warm = data["temp"] > 30 # eine Maske, die True für alle Zeilen mit Temperatur > 30 zurückgibt
display(data[maske_warm].head()) # die ersten 5 Zeilen mit Temperatur > 30 °C
print(maske_warm.sum(), "Stunden mit über 30 °C") # Anzahl der Stunden (Zeilen) mit Temperatur > 30 °C
```

Mehrere Bedingungen werden mit `&` (und) bzw. `|` (oder) verknüpft. Jede Bedingung muss dabei in **eigenen Klammern** stehen.

```python
warm_und_arbeitstag = data[(data["temp"] > 30) & (data["workingday"] == 1)]
print(len(warm_und_arbeitstag), "Stunden mit über 30 °C an Arbeitstagen")
```

## Neue Spalten berechnen
Rechnen mit Spalten ist **vektorisiert**, wirkt also auf alle Zeilen auf einmal.

```python
data["anteil_casual"] = data["casual"] / data["count"]
data[["casual", "count", "anteil_casual"]].head()
```

## Eine Funktion auf jeden Wert anwenden: `.apply`
Mit `.apply(funktion)` wendet pandas eine Funktion auf **jeden einzelnen Wert** einer Spalte an. Das Ergebnis ist eine neue Spalte mit genauso vielen Zeilen.

Beispiel: Wir teilen die Temperatur in drei Kategorien ein. Dafür schreiben wir eine Funktion, die **einen** Wert bekommt und **eine** Kategorie zurückgibt. pandas ruft sie dann für jede der 5422 Zeilen auf. Wir zeigen die ersten 20 Zeilen aus dem DataFrame.

```python
def temperatur_kategorie(t):
    if t < 10:
        return "kalt"
    elif t < 25:
        return "mild"
    else:
        return "warm"

data["temp_kategorie"] = data["temp"].apply(temperatur_kategorie)   
data[["temp", "temp_kategorie"]].head(20)
```

Die Auswahl lässt sich wie jeder andere Wert in einer **neuen Variable** speichern. `subset` ist dann eine eigene, kleinere Tabelle mit nur diesen drei Spalten, und `data` bleibt unverändert.

```python
subset = data[["temp", "humidity", "count"]]
subset
```

## Spalten und Zeilen entfernen
Mit `.drop(columns=[...])` entfernen wir Spalten. Wie viele pandas-Methoden gibt `.drop` eine **neue** Tabelle zurück. `data` selbst bleibt unverändert, solange wir das Ergebnis nicht wieder in `data` speichern.

```python
data_klein = data.drop(columns=["atemp", "casual", "registered"])

print(data.shape)        # Original
print(data_klein.shape)  # drei Spalten weniger
```

Zeilen entfernen heißt in pandas fast immer **filtern**: Wir behalten nur die Zeilen, die eine Bedingung erfüllen. `!=` bedeutet „ungleich“.

```python
ohne_regen = data[data["weather"] != 3]   # alle Stunden ohne Regen behalten
print(len(data), "->", len(ohne_regen), "Zeilen")
```

## Fehlende Werte und Duplikate
Vor jeder Analyse sollte man prüfen, ob die Daten „sauber“ sind. Zwei typische Probleme:

- **Fehlende Werte:** leere Zellen, in pandas als `NaN` („Not a Number“) angezeigt
- **Duplikate:** Zeilen, die komplett doppelt vorkommen, z. B. weil Daten zweimal importiert wurden

`.isna()` erzeugt eine Maske: `True` überall dort, wo ein Wert fehlt. Mit `.sum()` zählen wir die Treffer. `.duplicated()` markiert jede Zeile, die schon einmal vorkam.

```python
print(data.isna().sum().sum(), "fehlende Werte insgesamt")
print(data.duplicated().sum(), "doppelte Zeilen")
```

Damit Sie sehen, wie es aussieht, wenn nicht, bauen wir eine kleine Tabelle von Hand. Ein DataFrame lässt sich direkt aus einem **Dictionary aus Listen** erzeugen: ein Schlüssel pro Spalte.

```python
import numpy as np

messungen = pd.DataFrame({
    "station":   ["A",      "A",      "B",      "B",      "B"],
    "datum":     ["01.05.", "02.05.", "01.05.", "02.05.", "02.05."],
    "ausleihen": [120,      np.nan,   80,       95,       95],
})
messungen
```

```python
print(messungen.isna().sum())        # fehlende Werte pro Spalte
print(messungen.duplicated().sum())  # komplett doppelte Zeilen
```

```python
bereinigt = messungen.drop_duplicates()
bereinigt = messungen.dropna()   # erst Duplikate, dann Zeilen mit fehlenden Werten entfernen
bereinigt
```


## Union: Tabellen untereinander hängen
Bei einer **Union** werden Tabellen mit denselben Spalten **untereinander** gehängt, z. B. zwei Exporte aus demselben System. In pandas geht das mit `pd.concat([...])`. `ignore_index=True` nummeriert die Zeilen danach neu durch.

```python
export_montag = pd.DataFrame({"part_id": [367, 856], "batch": ["006", "009"], "quantity": [200, 550]})
export_dienstag = pd.DataFrame({"part_id": [856, 562], "batch": ["009", "002"], "quantity": [550, 830]})

alle_batches = pd.concat([export_montag, export_dienstag], ignore_index=True)
alle_batches
```

🤔 **Erst denken, dann ausführen:** Welches Datenproblem ist bei dieser Union entstanden, und wie beheben Sie es?

```python
print(alle_batches.duplicated().sum(), "doppelte Zeile(n)")
alle_batches.drop_duplicates()
```

## Join: Tabellen über einen Schlüssel verbinden
In Übung 1 haben Sie eine Tabelle in die 3. Normalform gebracht, also in mehrere kleine Tabellen **aufgeteilt**. Ein **Join** macht genau das Umgekehrte: Er **verbindet** Tabellen wieder über einen gemeinsamen Schlüssel. In pandas: `pd.merge(links, rechts, on="schlüssel", how=...)`.

Hier zwei Tabellen aus der Normalisierungsaufgabe. In der Werk-Tabelle fehlt absichtlich Münster:

```python
teile = pd.DataFrame({
    "part_id":      [367, 856, 562],
    "denomination": ["Lever", "Hand-wheel", "Screw"],
    "color":        ["silver", "black", "grey"],
    "zip_code":     ["33095", "33613", "48143"],
})
werke = pd.DataFrame({
    "zip_code": ["33095", "33613"],
    "facility": ["Paderborn", "Bielefeld"],
})
display(teile)
display(werke)
```

```python
pd.merge(teile, werke, on="zip_code", how="inner")   # nur Zeilen, deren Schlüssel in BEIDEN Tabellen vorkommt
```

```python
pd.merge(teile, werke, on="zip_code", how="outer")    # alle Zeilen aus der linken Tabelle
```

> 💡 **Konzept – Welcher Join?**
>
> | `how=` | Ergebnis |
> |---|---|
> | `"inner"` | nur Zeilen mit passendem Schlüssel in **beiden** Tabellen – die Schraube (48143) verschwindet |
> | `"outer"` | alle Zeilen aus beiden Tabellen |
>

# 2 Funktionen

## Wiederhole dich nicht
Angenommen, wir wollen Temperaturen von Celsius in Fahrenheit umrechnen, und zwar für viele Stunden. Die naheliegende Lösung ist Kopieren:

```python
temp_stunde_1 = 9.84
fahrenheit_stunde_1 = temp_stunde_1 * 9 / 5 + 32
print(fahrenheit_stunde_1)

temp_stunde_2 = 9.02
fahrenheit_stunde_2 = temp_stunde_2 * 9 / 5 + 32
print(fahrenheit_stunde_2)

temp_stunde_3 = 13.12
fahrenheit_stunde_3 = temp_stunde_3 * 9 / 5 + 32
print(fahrenheit_stunde_3)
```

Wenn Sie merken, dass Sie Code kopieren, ist das ein Signal: Hier gehört eine **Funktion** hin. Eine Funktion ist ein benannter, wiederverwendbarer Code-Baustein:

- `def` leitet die Definition ein, danach kommen der Name und in Klammern die **Parameter** (die Eingaben).
- Der Code der Funktion ist **eingerückt** (4 Leerzeichen). Die Einrückung ist in Python keine Kosmetik, sondern legt fest, was zur Funktion gehört.
- `return` gibt das Ergebnis zurück.

```python
def celsius_zu_fahrenheit(celsius):
    return celsius * 9 / 5 + 32

print(celsius_zu_fahrenheit(9.84))
print(celsius_zu_fahrenheit(9.02))
print(celsius_zu_fahrenheit(13.12))
```

Was passiert beim Aufruf `celsius_zu_fahrenheit(9.84)`?
1. Python springt in die Funktion und weist den Wert dem Parameter zu: `celsius = 9.84`.
2. Der eingerückte Code läuft: `9.84 * 9 / 5 + 32`, also `49.712`.
3. `return` gibt dieses Ergebnis **zurück** an die Stelle, an der die Funktion aufgerufen wurde, hier an `print`.

Der Parameter `celsius` ist eine Variable, die es nur *innerhalb* der Funktion gibt, als Platzhalter für den Wert, den wir beim Aufruf übergeben.



![Meme](https://raw.githubusercontent.com/kbrennig/MODS_WS26_27/main/Intro_to_Python/images/03_indentation.png)

## Funktionen mit mehreren Eingaben
Eine Funktion kann auch mehrere Parameter haben. Diese Funktion berechnet, welchen Anteil ein Teil an einem Ganzen hat:

```python
def anteil(teil, gesamt):
    ergebnis = teil / gesamt
    return ergebnis

anteil_registered = anteil(2, 8)
print(anteil_registered)
```


🤔 **Erst denken, dann ausführen:** Was kommt bei `anteil(50, 1000)` heraus?

```python
print(anteil(50, 1000))
```

## Benannte Argumente
Damit die Reihenfolge nicht zur Fehlerquelle wird, können wir beim Aufruf dazuschreiben, welcher Wert zu welchem Parameter gehört. Dann ist die Reihenfolge egal, und man sieht sofort, was was ist:

```python
print(anteil(teil=13, gesamt=16))
print(anteil(gesamt=16, teil=13))   
```



## Standardwerte
Ein Parameter kann einen **Standardwert** haben (`= ...` in der Definition). Gibt man beim Aufruf nichts an, wird der Standardwert benutzt; gibt man etwas an, wird er überschrieben. Hier runden wir das Ergebnis mit der eingebauten Funktion `round(zahl, stellen)`, standardmäßig auf 2 Nachkommastellen:

```python
def anteil(teil, gesamt, stellen=2):
    ergebnis = teil / gesamt
    return round(ergebnis, stellen)

print(anteil(13, 16))              # stellen nicht angegeben -> Standardwert 2
print(anteil(13, 16, stellen=4))   # Standardwert überschrieben
```

Deshalb kann man viele Funktionen mit nur ein, zwei Argumenten aufrufen, obwohl sie Dutzende Einstellungen haben: Alles, was man nicht angibt, steht auf einem sinnvollen Standardwert.

# 3 Listen und Dictionaries

## Listen: viele Werte in einer Reihenfolge
Für jede Stunde eine eigene Variable anzulegen, wird schnell unübersichtlich. Eine **Liste** speichert viele Werte in einer festen Reihenfolge, in eckigen Klammern und durch Kommas getrennt:

```python
counts = [16, 40, 32, 13, 1, 1, 2, 3]   # Ausleihen in den ersten 8 Stunden
print(counts)
```

Auf einzelne Elemente greifen wir mit eckigen Klammern und ihrer Position (dem **Index**) zu. **Achtung: Python fängt bei 0 an zu zählen!**

```python
print(counts[0])    # erstes Element
print(counts[1])    # zweites Element
print(counts[-1])   # letztes Element
```

![Meme](https://raw.githubusercontent.com/kbrennig/MODS_WS26_27/main/Intro_to_Python/images/04_zero_indexing.png)

Mit einem Doppelpunkt schneiden wir **Bereiche** aus (Slicing). `start:stop` bedeutet: von `start` bis **ausschließlich** `stop`. Ein Doppelpunkt allein heißt „alles“.

```python
print(counts[0:3])   # Index 0, 1, 2
print(counts[:3])    # dasselbe, Start weggelassen
print(counts[5:])    # ab Index 5 bis zum Ende
print(counts[:])     # alles
```

> 💡 **Konzept – Funktionen und Methoden**
>
> Es gibt zwei Schreibweisen, die Sie ständig sehen werden:
> - **Funktion:** `funktion(objekt)` – z. B. `len(x)`, `print(x)`
> - **Methode:** `objekt.methode()` – eine Funktion, die zu einem bestimmten Objekt *gehört*, z. B. `counts.append(8)`
> 
> Lesen Sie den Punkt als „**mach mit diesem Objekt …**“. Welche Methoden es gibt, hängt vom Typ des Objekts ab.

```python
# Funktionen len(), sum(), max(), min() auf Listen
print(len(counts))
print(sum(counts))
print(max(counts))
print(min(counts))

# Methoden auf Listen
print(counts)
counts.append(8)     
print(counts)
```

## Dictionaries: Werte über Namen nachschlagen
Ein **Dictionary** (`dict`) speichert Paare aus **Schlüssel** und **Wert**, in geschweiften Klammern. Statt über eine Position greifen wir über den Schlüssel zu, wie in einem Wörterbuch.

```python
stunde = {"temp": 9.84, "humidity": 81, "count": 16}
print(stunde["temp"])

jahreszeiten = {1: "Frühling", 2: "Sommer", 3: "Herbst", 4: "Winter"}
print(jahreszeiten[3])
```

> 💡 **Konzept – Liste oder Dictionary?**
>
> - **Liste:** Die *Reihenfolge* zählt, Zugriff über die Position (`counts[0]`).
> - **Dictionary:** Die *Bedeutung* zählt, Zugriff über einen Namen (`stunde["temp"]`).
> 
> Ein Datensatz ist eigentlich beides: Jede Spalte ist eine Liste von Werten, und die Spalten finden wir über ihren Namen. Genau so ist eine pandas-DataFrame (Teil 6) aufgebaut.

# 4 Entscheiden und Wiederholen

## if / elif / else
Manchmal soll Code nur unter einer Bedingung laufen. Nach `if` steht ein Ausdruck, der `True` oder `False` ergibt. Der eingerückte Block darunter läuft nur, wenn er `True` ist. Mit `elif` (else if) prüfen wir weitere Bedingungen, `else` fängt alles Übrige ab.

Im BikeSharing-Datensatz ist das Wetter als Zahl kodiert:

```python
weather = 2

if weather == 1:
    print("klar")
elif weather == 2:
    print("neblig/bewölkt")
elif weather == 3:
    print("leichter Regen/Schnee")
else:
    print("starker Regen/Schnee")
```

## for-Schleifen
Eine `for`-Schleife führt denselben Code **für jedes Element** einer Sammlung aus. Die Variable nach `for` zeigt nacheinander auf jedes Element.

```python
temps = [9.84, 9.02, 9.02, 9.84, 9.84, 9.84, 9.02, 8.2]

for t in temps:
    print(t, "°C =", celsius_zu_fahrenheit(t), "°F")
```

Typisches Muster: eine leere Liste anlegen und in der Schleife Ergebnisse anhängen.

```python
temps_f = []
for t in temps:
    temps_f.append(celsius_zu_fahrenheit(t))
print(temps_f)
```

## List Comprehensions: dasselbe in einer Zeile
Weil dieses Muster so häufig ist, hat Python eine Kurzschreibweise dafür. Lesen Sie sie wie einen Satz: *„Gib mir `celsius_zu_fahrenheit(t)` für jedes `t` in `temps`.“*

```python
temps_f = [celsius_zu_fahrenheit(t) for t in temps]
print(temps_f)
```

Mit einem angehängten `if` wird zusätzlich **gefiltert**: *„… aber nur, wenn die Bedingung erfüllt ist.“*

```python
counts = [16, 40, 32, 13, 1, 1, 2, 3, 8]
viel_los = [c for c in counts if c > 10]
print(viel_los)
```

## Vektorisierung: eine Operation auf alle Werte
NumPys zentrale Datenstruktur ist das **Array**. Anders als bei einer Liste wirkt eine Rechenoperation auf einem Array auf **jedes Element gleichzeitig**. Eine Schleife ist nicht nötig => schneller

```python
temps = [9.84, 9.02, 9.02, 9.84, 9.84, 9.84, 9.02, 8.2]

temps_array = np.array(temps)
print(temps_array)
```

Vektorisierung ist **viel** schneller. Ein Vergleich mit 10 Millionen Werten (dauert ein paar Sekunden):

```python
import time

n = 10_000_000
viele_temps = list(np.random.uniform(-5, 35, n))
viele_temps_array = np.array(viele_temps)

start = time.time()
ergebnis_schleife = [t * 9 / 5 + 32 for t in viele_temps]
print("Schleife:     ", round(time.time() - start, 2), "Sekunden")

start = time.time()
ergebnis_array = viele_temps_array * 9 / 5 + 32
print("Vektorisiert: ", round(time.time() - start, 2), "Sekunden")
```

![Meme](https://raw.githubusercontent.com/kbrennig/MODS_WS26_27/main/Intro_to_Python/images/07_vectorized.png)

Wie bei Listen gibt es für Arrays statistische Auswertungen, hier als **Methoden** direkt auf dem Array:

```python
print(temps_array.mean())   # Mittelwert
print(temps_array.std())    # Standardabweichung
print(temps_array.min(), temps_array.max())
```

## Bedingungen auf ganze Arrays: Masken
Auch Vergleiche funktionieren elementweise. Heraus kommt ein Array aus `True`/`False`, eine sogenannte **Maske**:

```python
counts_array = np.array([16, 40, 32, 13, 1, 1, 2, 3, 8])
maske = counts_array > 10
print(maske)
print(counts_array[maske])   # nur die Werte, bei denen die Maske True ist
print(maske.sum())           # True zählt als 1 -> Anzahl der Treffer
```

## Zeilen filtern
Hier kommen die Masken aus Teil 5 zurück: Eine Bedingung auf eine Spalte ergibt eine Maske, und mit der Maske wählen wir Zeilen aus.

```python
display(data.iloc[0])   # erste Zeile

maske_warm = data["temp"] > 30 # eine Maske, die True für alle Zeilen mit Temperatur > 30 zurückgibt
display(data[maske_warm].head()) # die ersten 5 Zeilen mit Temperatur > 30 °C
print(maske_warm.sum(), "Stunden mit über 30 °C") # Anzahl der Stunden (Zeilen) mit Temperatur > 30 °C
```

Mehrere Bedingungen werden mit `&` (und) bzw. `|` (oder) verknüpft. Jede Bedingung muss dabei in **eigenen Klammern** stehen.

```python
warm_und_arbeitstag = data[(data["temp"] > 30) & (data["workingday"] == 1)]
print(len(warm_und_arbeitstag), "Stunden mit über 30 °C an Arbeitstagen")
```

## Eine Funktion auf jeden Wert anwenden: `.apply`
Mit `.apply(funktion)` wendet pandas eine Funktion auf **jeden einzelnen Wert** einer Spalte an. Das Ergebnis ist eine neue Spalte mit genauso vielen Zeilen.

Beispiel: Wir teilen die Temperatur in drei Kategorien ein. Dafür schreiben wir eine Funktion, die **einen** Wert bekommt und **eine** Kategorie zurückgibt. pandas ruft sie dann für jede der 5422 Zeilen auf. Wir zeigen die ersten 20 Zeilen aus dem DataFrame.

```python
def temperatur_kategorie(t):
    if t < 10:
        return "kalt"
    elif t < 25:
        return "mild"
    else:
        return "warm"

data["temp_kategorie"] = data["temp"].apply(temperatur_kategorie)   
data[["temp", "temp_kategorie"]].head(20)
```

Die Auswahl lässt sich wie jeder andere Wert in einer **neuen Variable** speichern. `subset` ist dann eine eigene, kleinere Tabelle mit nur diesen drei Spalten, und `data` bleibt unverändert.

```python
subset = data[["temp", "humidity", "count"]]
subset
```

# Lineare Regression von Hand

Wählt den Achsenabschnitt (Intercept) $\beta_0$ und die Steigung $\beta_1$ selbst und beobachtet, wie sich die Summe der quadrierten Residuen (RSS) verändert.

$$e_i = y_i - \hat{y}_i \qquad RSS = e_1^2 + e_2^2 + \dots + e_n^2$$

Die Methode der kleinsten Quadrate (Ordinary Least Squares, OLS) wählt $\beta_0$ und $\beta_1$ so, dass die RSS möglichst klein wird. Schafft ihr es, nah an die OLS-Gerade heranzukommen?

# Arbeitsumgebung einrichten

Pakete importieren.

```python
import pandas as pd
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
from statsmodels.formula.api import ols
```

# Daten laden

Den Advertising-Datensatz laden (alle 200 Märkte).

```python
path = "https://raw.githubusercontent.com/kbrennig/MODS_WS25_26/refs/heads/main/data/Advertising.csv"
df = pd.read_csv(path)
df.head()
```

# Die Gerade von Hand anpassen

Bewegt die Regler. Die grauen Linien sind die Residuen $e_i$, im Titel seht ihr eure RSS.

Tipp: Der Plot aktualisiert sich, sobald ihr den Regler loslasst. Zum Feintuning einen Regler anklicken und die Pfeiltasten (← →) benutzen.

```python
# OLS solution, used only for comparison (minimum possible RSS)
ols_model = ols(formula="Sales ~ TV", data=df).fit()
RSS_OLS = ols_model.ssr

def plot_line(beta0, beta1, show_ols=False):
    # Calculate the prediction and the residuals e_i = y_i - y_hat_i
    prediction = beta0 + beta1 * df["TV"]
    residual = df["Sales"] - prediction

    # Calculate the RSS
    RSS = (residual**2).sum()

    # Plot data, residuals and the chosen line
    fig, ax = plt.subplots(figsize=(8, 5))
    ax.vlines(df["TV"], prediction, df["Sales"], color="grey", alpha=0.4, linewidth=1)
    ax.scatter(df["TV"], df["Sales"])
    ax.plot(df["TV"], prediction, color="red", label="your line")
    if show_ols:
        ax.plot(df["TV"], ols_model.predict(df), color="black", linestyle="--",
                label=f"OLS: {ols_model.params['Intercept']:.4f} + {ols_model.params['TV']:.4f} × TV")
        ax.legend()
    ax.set_ylim(-10, 40)  # fix limits of the y-axis
    ax.set_xlabel("TV")
    ax.set_ylabel("Sales")
    fig.suptitle(f"Sales ≈ {beta0:.2f} + {beta1:.3f} × TV")
    ax.set_title(f"RSS: {RSS:.2f}   (minimum with OLS: {RSS_OLS:.2f})")
    plt.show()

interact(
    plot_line,
    beta0=FloatSlider(value=12.0, min=-5.0, max=20.0, step=0.1,
                      continuous_update=False, description="beta 0"),
    beta1=FloatSlider(value=-0.1, min=-0.2, max=0.2, step=0.001, readout_format=".3f",
                      continuous_update=False, description="beta 1"),
    show_ols=False,
);
```
